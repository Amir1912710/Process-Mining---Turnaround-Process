```python
from FlightRadar24 import FlightRadar24API

fr_api = FlightRadar24API()

# Get airport details
airport = fr_api.get_airport("BUD")

# Get arrivals and departures using the dedicated methods
arrivals = fr_api.get_airport_arrivals(airport["iata"])
departures = fr_api.get_airport_departures(airport["iata"])

print("Arrivals:")
for flight in arrivals[:5]:
    print(f"{flight['flight']['identification']['callsign']} from {flight['flight']['airport']['origin']['name']}")

print("\nDepartures:")
for flight in departures[:5]:
    print(f"{flight['flight']['identification']['callsign']} to {flight['flight']['airport']['destination']['name']}")
```
