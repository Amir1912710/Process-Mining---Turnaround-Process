# 04 ACDM App and Airline Segment Comparison

This notebook loads manually captured ACDM app observations, builds an app milestone event log, compares low-cost and full-service carriers, and prepares simulator-ready profile tables.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.load_acdm_app_data import load_acdm_app_data
from src.build_acdm_app_event_log import build_acdm_app_event_log
from src.compare_airline_segments import compare_airline_segments
from src.build_simulator_inputs import build_simulator_inputs
from src.config import get_data_paths, load_config
from src.utils import read_table
```

```python
config_path = PROJECT_ROOT / 'configs' / 'default_config.yaml'
segment_config = PROJECT_ROOT / 'configs' / 'airline_segments.yaml'
config = load_config(config_path)
paths = get_data_paths(config, PROJECT_ROOT)

load_acdm_app_data(config_path=config_path, segment_config_path=segment_config)
build_acdm_app_event_log(config_path=config_path)
compare_airline_segments(config_path=config_path, segment_config_path=segment_config)
build_simulator_inputs(config_path=config_path, segment_config_path=segment_config)
```

```python
app_df = read_table(paths['interim_dir'] / 'acdm_app_snapshots.parquet')
app_event_log = read_table(paths['processed_dir'] / 'acdm_app_event_log.parquet')
sim_profiles = read_table(paths['processed_dir'] / 'simulator_turnaround_profiles.parquet')
print({'app_snapshots': len(app_df), 'app_event_rows': len(app_event_log), 'simulator_profiles': len(sim_profiles)})
app_df.head()
```
